```python
import json
import os
import uuid

from langchain_experimental.text_splitter import SemanticChunker
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def create_semantic_chunks(structured_doc: dict) -> list:
    """
    Bước 1: Tạo các chunk giàu ngữ nghĩa từ cấu trúc JSON.
    """
    metadata = {key: value for key, value in structured_doc.items() if key != "chuong"}
    content_tree = structured_doc.get("chuong", [])
    semantic_chunks = []

    for chapter in content_tree:
        # Bắt đầu với đường dẫn ngữ cảnh là tên văn bản
        _recursive_chunker(
            node=chapter,
            context_path=[metadata.get("ten_van_ban", metadata.get("file_name"))],
            citation_path=[],
            metadata=metadata,
            chunks_list=semantic_chunks
        )
    return semantic_chunks
```

```python
def _recursive_chunker(node: dict, context_path: list, citation_path: list, metadata: dict, chunks_list: list):
    """Hàm đệ quy để duyệt cây JSON và tạo chunk."""
    current_id, current_title = "", ""
    if "ten_chuong" in node and node["ten_chuong"]:
        current_id, current_title = node["ten_chuong"], node["tieu_de"]
    elif "ten_muc" in node and node["ten_muc"]:
        current_id, current_title = node["ten_muc"], node["tieu_de"]
    elif "ten_khoan" in node and node["ten_khoan"]:
        current_id, current_title = node["ten_khoan"], node["tieu_de"]

    new_context_path = context_path + [current_title] if current_title else context_path
    new_citation_path = citation_path + [current_id] if current_id else citation_path

    children_keys = ["muc", "khoan", "diem"]
    found_children = False

    for key in children_keys:
        if key in node and node[key]:
            found_children = True
            if key == "diem":
                for point_content in node[key]:
                    final_citation = ", ".join(new_citation_path)
                    enriched_content = (
                        f"Văn bản: {metadata.get('ten_van_ban', 'N/A')}. "
                        f"Ngày ban hành: {metadata.get('ngay_ban_hanh', 'N/A')}. "
                        f"Người ký: {metadata.get('nguoi_ky', 'N/A')}, {metadata.get('chuc_vu_nguoi_ky', 'N/A')}. "
                        f"Nội dung: {'. '.join(new_context_path)}. {point_content}"
                    )
                    chunks_list.append({
                        "id": str(uuid.uuid4()),
                        "source": metadata.get("ten_van_ban", "N/A"),  # Đổi thành ten_van_ban
                        "location": final_citation,
                        "content": enriched_content.strip(),
                        "issue_date": metadata.get("ngay_ban_hanh", "N/A"),
                        "effective_date": metadata.get("ngay_hieu_luc", "N/A"),
                        "signatory": metadata.get("nguoi_ky", "N/A"),
                        "signatory_position": metadata.get("chuc_vu_nguoi_ky", "N/A"),
                    })
            else:
                for child_node in node[key]:
                    _recursive_chunker(child_node, new_context_path, new_citation_path, metadata, chunks_list)

    if not found_children and current_title:
        final_citation = ", ".join(new_citation_path)
        enriched_content = (
            f"Văn bản: {metadata.get('ten_van_ban', 'N/A')}. "
            f"Ngày ban hành: {metadata.get('ngay_ban_hanh', 'N/A')}. "
            f"Người ký: {metadata.get('nguoi_ky', 'N/A')}, {metadata.get('chuc_vu_nguoi_ky', 'N/A')}. "
            f"Nội dung: {'. '.join(new_context_path)}"
        )
        chunks_list.append({
            "id": str(uuid.uuid4()),
            "source": metadata.get("ten_van_ban", "N/A"),
            "location": final_citation,
            "content": enriched_content.strip(),
            "issue_date": metadata.get("ngay_ban_hanh", "N/A"),
            "effective_date": metadata.get("ngay_hieu_luc", "N/A"),
            "signatory": metadata.get("nguoi_ky", "N/A"),
            "signatory_position": metadata.get("chuc_vu_nguoi_ky", "N/A"),
        })
```

```python
def refine_chunks_semantically(semantic_chunks: list, semantic_splitter, char_splitter) -> list:
    refined_chunks = []
    print("\n--- [Bước 2] Bắt đầu tinh chỉnh kích thước chunk... ---")
    for chunk in semantic_chunks:
        content = chunk['content']
        if len(content) < 1500:  # Tăng ngưỡng
            refined_chunks.append(chunk)
        else:
            # Hybrid: Dùng char splitter trước, rồi semantic nếu cần
            sub_contents = char_splitter.split_text(content)
            for sub in sub_contents:
                if len(sub) > 1500:
                    semantic_subs = semantic_splitter.split_text(sub)
                    # Chỉ thêm "(phần)" nếu có nhiều semantic_subs
                    if len(semantic_subs) > 1:
                        for i, semantic_sub in enumerate(semantic_subs):
                            refined_chunks.append({
                                "id": str(uuid.uuid4()),
                                "source": chunk['source'],
                                "location": f"{chunk['location']} (phần {i + 1})",  # Giữ nếu tách nhiều
                                "content": semantic_sub,
                                "issue_date": chunk['issue_date'],
                                "effective_date": chunk['effective_date'],
                                "signatory": chunk['signatory'],
                                "signatory_position": chunk['signatory_position'],
                            })
                    else:
                        # Nếu chỉ 1 sub, không thêm "(phần)"
                        refined_chunks.append({
                            "id": str(uuid.uuid4()),
                            "source": chunk['source'],
                            "location": chunk['location'],
                            "content": semantic_subs[0],
                            "issue_date": chunk['issue_date'],
                            "effective_date": chunk['effective_date'],
                            "signatory": chunk['signatory'],
                            "signatory_position": chunk['signatory_position'],
                        })
                else:
                    refined_chunks.append({
                        "id": str(uuid.uuid4()),
                        "source": chunk['source'],
                        "location": chunk['location'],  # Không thêm "(phần)"
                        "content": sub,
                        "issue_date": chunk['issue_date'],
                        "effective_date": chunk['effective_date'],
                        "signatory": chunk['signatory'],
                        "signatory_position": chunk['signatory_position'],
                    })
    print(f"--- [Bước 2] Hoàn thành, tổng số chunk: {len(refined_chunks)} ---")
    return refined_chunks
```

```python
print(">>> Đang khởi tạo mô hình embedding...")
# embedding_model = GoogleGenerativeAIEmbeddings(
#     model="models/text-embedding-004",
#     google_api_key="AIzaSyDxexrY4I2jRslG8Ig17-mNBfP1AeAO5aY"
# )

import re

print(f">>> Đang tải cơ sở dữ liệu vector từ: {vectorstore_path}...")
vectorstore = Chroma(
    persist_directory=vectorstore_path,
    embedding_function=embedding_model
)

# Function parse source và location từ query
def parse_query_for_filter(query):
    # Parse source (giữ nguyên, ok rồi)
    source_match = re.search(r"(\d+/CT-[A-Z]+)", query, re.IGNORECASE)
    source = f"Chỉ Thị {source_match.group(1)}" if source_match else None

    # Regex refine cho location: Match "Chương X, mục Y, khoản Z" hoặc partial, có dấu, ignore case
    location_match = re.search(r"(chương\s*[ivxlcdm0-9]+)?\s*,?\s*(mục\s*\d+)?\s*,?\s*(khoản\s*[a-z0-9]+)?", query, re.IGNORECASE)
    location_parts = [part.strip() for part in location_match.groups() if part] if location_match else []
    location = ', '.join(location_parts) if location_parts else None  # Join thành string như "Mục 1, Khoản a"

    return source, location


from langchain_core.documents import Document  # Import nếu chưa có
from rank_bm25 import BM25Okapi

# Hàm helper để loại duplicate Document (dựa trên content + id, tránh unhashable)
def unique_documents(docs):
    seen = set()
    unique_docs = []
    for doc in docs:
        # Identifier unique: page_content + id (nếu có)
        identifier = (doc.page_content, doc.metadata.get('id', ''))
        if identifier not in seen:
            seen.add(identifier)
            unique_docs.append(doc)
    return unique_docs

# Retriever với dynamic filter + hybrid search (sửa unique để tránh TypeError)
def filtered_retriever(query):
    source, location = parse_query_for_filter(query)

    conditions = []
    if source:
        conditions.append({"source": {"$eq": source}})

    filter_dict = {"$and": conditions} if len(conditions) > 1 else conditions[0] if conditions else {}

    k = 10 if len(query.split()) > 10 else 5

    search_kwargs = {"k": k, "filter": filter_dict} if filter_dict else {"k": k}
    embedding_results = vectorstore.as_retriever(search_kwargs=search_kwargs).invoke(query)  # Embedding search

    if source:  # Hybrid: Thêm BM25 keyword search trên all chunks của source
        data = vectorstore.get(where={"source": {"$eq": source}})
        all_docs = data['documents']
        all_metadatas = data['metadatas']

        combined = [Document(page_content=doc, metadata=meta) for doc, meta in zip(all_docs, all_metadatas)]

        tokenized_docs = [doc.page_content.split() for doc in combined]
        bm25 = BM25Okapi(tokenized_docs)
        tokenized_query = query.split()
        bm25_scores = bm25.get_scores(tokenized_query)

        bm25_results = [combined[i] for i in sorted(range(len(bm25_scores)), key=lambda j: bm25_scores[j], reverse=True)[:k]]

        # Kết hợp và loại duplicate dùng hàm helper (fix TypeError)
        unique_results = unique_documents(embedding_results + bm25_results)

        # Post-filter cho location nếu có
        if location:
            unique_results = [doc for doc in unique_results if location.lower() in doc.metadata.get('location', '').lower()]
    else:
        unique_results = embedding_results

    if not unique_results:
        print(f"Error: Không tìm thấy chunks cho query '{query}'.")

    return unique_results


print(">>> Filtered Retriever refined (loose match) đã sẵn sàng.")
```
